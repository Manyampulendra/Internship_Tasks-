## Q1. Find the Shortest Word in a Sentence

### 1. Observe
Sentence is given; output the shortest word.

### 2. Understand
Identify the word with the fewest characters.

### 3. Break
Split into words, compare lengths, and keep the first word in a tie.

### 4. Logic
Split the sentence into words and select the shortest. Keep the first word if lengths tie.

### 5. Code
Run the next code cell after reading the approach.

### 6. Test
The cell tests the supplied example and one additional input. Check the outputs to verify the logic.

**Edge cases:** Empty sentences return `None` for word-selection tasks, empty lists return `None` for largest/smallest tasks, and counting/dictionary tasks return zero or empty collections when appropriate.


```python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        return None
    answer = words[0]
    for word in words[1:]:
        if len(word) < len(answer):
            answer = word
    return answer

print(shortest_word("Python makes coding easy"))  # easy
print(shortest_word("we learn Python"))           # we
```

### 4. Logic

Run the next code cell after reading the approach.

```python
def most_frequent_character(text):
    text = text.replace(" ", "")
    if not text:
        return None
    counts = {}
    for char in text:
        counts[char] = counts.get(char, 0) + 1
    answer = text[0]
    for char in counts:
        if counts[char] > counts[answer]:
            answer = char
    return answer

print(most_frequent_character("banana"))  # a
print(most_frequent_character("aabb"))    # a
```

## Q16. Find the Most Frequent Word in a Sentence

### 1. Observe
A sentence is given; capitalization should not matter.

### 2. Understand
Find the most frequent word, with first-seen tie-breaking.

### 3. Break
Lowercase words, count them, then scan in sentence order.

### 4. Logic
Lowercase words, count occurrences, and return the earliest word among tied maximum frequencies.

### 5. Code
Run the next code cell after reading the approach.

### 6. Test
The cell tests the supplied example and one additional input. Check the outputs to verify the logic.

**Edge cases:** Empty sentences return `None` for word-selection tasks, empty lists return `None` for largest/smallest tasks, and counting/dictionary tasks return zero or empty collections when appropriate.


```python
def most_frequent_word(sentence):
    words = sentence.lower().split()
    if not words:
        return None
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    answer = words[0]
    for word in words:
        if counts[word] > counts[answer]:
            answer = word
    return answer

print(most_frequent_word("Python is easy and Python is powerful"))  # python
print(most_frequent_word("Red blue red BLUE green"))                # red
```

## Q20. Find the Longest Word That Appears Only Once

### 1. Observe
A sentence is given; capitalization should not matter.

### 2. Understand
Find the longest word that occurs once; first wins ties.

### 3. Break
Count lowercase words and track the longest unique word.

### 4. Logic
Count lowercase words and track the longest word with frequency one. Use > to preserve the first tie.

### 5. Code
Run the next code cell after reading the approach.

### 6. Test
The cell tests the supplied example and one additional input. Check the outputs to verify the logic.

**Edge cases:** Empty sentences return `None` for word-selection tasks, empty lists return `None` for largest/smallest tasks, and counting/dictionary tasks return zero or empty collections when appropriate.


```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    answer = None
    for word in words:
        if counts[word] == 1:
            if answer is None or len(word) > len(answer):
                answer = word
    return answer

print(longest_unique_word("Python makes coding useful Python"))  # coding
print(longest_unique_word("cat dog bird fish"))                   # bird
```
