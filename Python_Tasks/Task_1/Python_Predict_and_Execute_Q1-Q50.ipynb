## Q35

**Prediction:** Output: [[1, 99], [3, 4]]

**Explanation:** A shallow copy shares nested lists; changing an inner list through b affects a.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

## Q36

**Prediction:** Output: [[1, 2], [3, 4]]

**Explanation:** A deep copy recursively copies nested lists, isolating them from changes to b.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```
